```python
import nbformat
from nbformat.v4 import new_notebook, new_code_cell, new_markdown_cell

def convert_custom_py_to_ipynb(input_py, output_ipynb):
    # Legge il file .py caricato
    with open(input_py, 'r', encoding='utf-8') as f:
        content = f.read()

    # Separa il contenuto basandosi sui tripli doppi apici (""")
    # In un file esportato da Colab, questo divide perfettamente codice e markdown
    blocks = content.split('"""')

    nb = new_notebook()

    for i, block in enumerate(blocks):
        # Evita di creare celle completamente vuote
        if not block.strip():
            continue

        # La logica dello split fa sì che i blocchi ad indice pari (0, 2, 4...) siano codice,
        # mentre quelli ad indice dispari (1, 3, 5...) racchiusi tra i """ siano Markdown.
        if i % 2 == 0:
            # Crea una cella di codice rimuovendo gli a capo iniziali/finali in eccesso
            nb.cells.append(new_code_cell(block.strip('\n')))
        else:
            # Crea una cella Markdown
            nb.cells.append(new_markdown_cell(block.strip('\n')))

    # Salva il nuovo Jupyter Notebook
    with open(output_ipynb, 'w', encoding='utf-8') as f:
        nbformat.write(nb, f)

    print(f"Conversione completata con successo! Il notebook è stato salvato come: {output_ipynb}")

# --- ESECUZIONE ---
# Sostituisci i nomi dei file se li hai rinominati nel tuo computer
file_ingresso = 'Jonathan_1.py'
file_uscita = 'Jonathan_1.ipynb'

convert_custom_py_to_ipynb(file_ingresso, file_uscita)
```
